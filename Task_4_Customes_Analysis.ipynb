```python
import pandas as pd
import numpy as np

print("="*60)
print("المشروع التكاملي: تنظيف ودمج وتحليل بيانات المتجر")
print("="*60)

# 1. إعداد جدول العملاء (8 عملاء مع قيم مفقودة وتكرار)
customers_df = pd.DataFrame({
    'customer_id': [1, 2, 3, 4, 5, 6, 7, 2],  # العميل 2 مكرر
    'name': ['Ahmad', 'Sara', 'Mahmoud', 'Layan', 'Omar', 'Raghad', 'Khalid', 'Sara'],
    'city': ['Amman', None, 'Irbid', 'Amman', None, 'Zarqa', 'Amman', None]
})

# 2. إعداد جدول الطلبات (15 طلب موزعة على العملاء)
orders_df = pd.DataFrame({
    'order_id': list(range(101, 116)),
    'customer_id': [1, 1, 2, 3, 3, 3, 5, 5, 6, 1, 7, 7, 7, 7, 1], # العميل 4 ليس لديه طلبات
    'amount': [120, 250, 400, 50, 90, 110, 300, 150, 210, 80, 500, 150, 200, 100, 310]
})

print("\n--- الخطوة 1 & 2: معاينة البيانات قبل التنظيف ---")
print("العملاء الأصليين:\n", customers_df)

# 3. تنظيف الجدولين
# معالجة القيم المفقودة في المدن
customers_df['city'] = customers_df['city'].fillna('Not Defined')

# حذف الصفوف المكررة
customers_cleaned = customers_df.drop_duplicates(keep='first')

print("\n--- الخطوة 3: جدول العملاء بعد التنظيف ---")
print(customers_cleaned)

# 4. دمج الجدولين بـ Left Merge
merged_df = pd.merge(customers_cleaned, orders_df, on='customer_id', how='left')

print("\n--- الخطوة 4: الجدول المدمج (Left Merge) ---")
print(merged_df.head(10))

# 5. باستخدام GroupBy: حساب عدد الطلبات وإجمالي الصرف
customer_summary = merged_df.groupby(['customer_id', 'name', 'city']).agg(
    order_count=('order_id', 'count'),
    total_spent=('amount', 'sum')
).reset_index()

# 6. ترتيب النتيجة تنازلياً وطباعة Top Customers
top_customers = customer_summary.sort_values(by='total_spent', ascending=False)

print("\n--- الخطوة 5 & 6: ملخص أداء العملاء مرتب تنازلياً (Top Customers) ---")
print(top_customers.to_string(index=False))

print("\n" + "="*60)
print("أفضل 3 عملاء حسب إجمالي الصرف (Top 3):  ")
print("="*60)
print(top_customers.head(3)[['name', 'city', 'order_count', 'total_spent']].to_string(index=False))



# خلاصة تحليل عملاء المتجر (Key Analytical Takeaway):
# أظهر التحليل بعد تنظيف البيانات ودمجها أن العميل "خالد" من مدينة عمان هو الأعلى إنفاقاً في المتجر بـ 950 $ عبر 4 طلبات
# يليه العميل "أحمد" من عمان أيضاً بإجمالي 760 $  كما تبيّن وجود تفاوت واضح في نشاط العملاء؛
# حيث إن العميل "ليان" لم تقم بأي عملية شراء على الإطلاق (0 طلبات وصرف 0 $)، بالإضافة إلى أن عملاء مدينة عمان يشكلون القوة الشرائية الضاربة للمتجر
# مما يوصي بتوجيه حملات تسويقية لإعادة تنشيط العملاء الخاملين مثل "ليان" وتركيز الحملات الترويجية على العاصمة عمان


```
