```python
# Plot Customization
import matplotlib.pyplot as plt
import numpy as np

# Now to customise two lines
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([13, 25, 30, 20])
y2 = np.array([17, 23, 38, 5])

line_style = dict(marker = ".",
                  markersize=30,
                  markerfacecolor="#1cd3fc",
                  markeredgecolor="#989c24",
                  linestyle="solid",
                  linewidth=4,
                  color="#1c5bfc")

# Use ** to unpack the line_style dictionary

plt.plot(x, y1, **line_style)
plt.plot(x, y2, **line_style)

plt.show()
```

```python
# Lbels
import matplotlib.pyplot as plt
import numpy as np

x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([13, 25, 30, 20])
y2 = np.array([17, 23, 38, 5])
y3 = np.array([13, 15, 20, 30])

plt.title("Class Size",
          fontsize=25,
          family="serif",
          fontweight="bold",
          color="#2d4cfc",) # to put a title for your chart, and to change the font size and font family and font weight and color for this title

plt.xlabel("Year", fontsize=20,
                   family="serif",
                   fontweight="bold",
                   color="blue") # this is the label for x axis

plt.ylabel("Students", fontsize=20,
                   family="serif",
                   fontweight="bold",
                   color="blue") # this is the label for y axis

plt.tick_params(axis="both",  # means tick parameters we use this to customise the horizintal and vertical values you can chose either x or y or both inside axis=" "
                colors="navy") # we used colors here not color because we chose both
plt.plot(x, y1)
plt.plot(x, y2)
plt.plot(x, y3)

plt.xticks(x) # to force pyhton to use the numbers in your x array

plt.show()
```

```python
# Pie Chart
import matplotlib.pyplot as plt
import numpy as np
# Pie Chart = Circular chart divided into slices to shoew percentege of the total, Good for visualizing distribution among categories.

categories = ["Freshmen", "Sophomores", "Juniors", "Seniors"] # here we kept it a list because no benefit to change it to numpy array
values = np.array([300, 250, 275, 225]) # here we changed the list to numpy array because its good when you deal with numbers
colors = ["red", "yellow", "blue", "green"] # we added this list to give a color for each category, you should add this list to the argument code

# plt.pie(values), if we use the code like this we will not have a labels on the pie chart
plt.pie(values, labels=categories,  # we already have the values we added categories as a labels because it shows labels names
        autopct="%1.1f%%",
        colors=colors,
        explode=[0, 0, 0, 0.1],
        shadow=True,
        startangle=90)   # autopct means auto percentege we used it here to show the percentege on each part of pie to display one digit after the dicimal
                         # use %1.1f instead of %1f
                         # to use percent sign use %1.1f%%
# we can explode any category using slices, we have four categories so we should have 4 values in the list we put it all zeros except for Seniors the futher number
# equal to the futher distance
# we can use shadow to drop a shadow for the pie
# we use startangle to rotate the pie, you could use for example 180 or 90


plt.title("Ehab Code College")

plt.show()
```

```python
# Scatter Plots
import matplotlib.pyplot as plt
import numpy as np
# scatter graph  = shows the relationship between two variables, Helps to identify correlation (+(Positive), -(Negative), None(No Correlation))
# Example: Study hours vs. Test scores \ that is a positive correlation

x1 = np.array([0, 1, 1, 2, 3, 4, 5, 6, 7, 7, 8]) # Hours Studied for the First Class
y1 = np.array([50, 60, 65, 62, 68, 70, 75, 78, 82, 85, 87]) # Grades for the First Class

# you can notice from the graph when the studies hours are more you are getting better grade and that is a positive correlation

x2 = np.array([0, 1, 2, 2, 3, 4, 5, 6, 7, 8, 8]) # Hours Studied for the Second Class
y2 = np.array([50, 58, 65, 70, 72, 78, 83, 88, 92, 95, 97]) # Grades for the Second Class

plt.scatter(x1, y1, color="skyblue",
            alpha=0.5,
            s=200,
            label="Class A")
# we used color here to change the scatter (dots) colors & alpha for transperency for example 0.5 means 50% and 0.1 means 10% & s for size
# Also we used legend (putting label for the class so we can read the grapgh better), but then you should call the legend function to make the labels visible

plt.scatter(x2, y2, color="red",
            alpha=0.5,
            s=200,
            label="Class B")

# put labels to make it more clear
plt.title("Test Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Grade")

plt.legend() # to make the labels visible
# the blue scattered dots is for class 1 (A), the red scattered dots is for class 2 (B)
plt.show()
```

```python
# Histograms
import matplotlib.pyplot as plt
import numpy as np
# Histogram  = A visual representation of the distribution of quantitative data. They gropu values into bins (intervals), and counts how many fall in each range
# its good for statistics and probabilities

scores = np.random.normal(loc=80, scale=10, size=100)
# we used 80 for the loc to make the scores aruond 80 % and most of the students are doing well but if we put it 50 it means most of the students around 50% and failed
# and we use scale for the standard diviattion the spread how far away are the scores going to diviate form the center the higher the number the greater the diviation
# we use size to define how many numbers we want to generate
scores = np.clip(scores, 0, 100)
# this is to force the numbers to stay between 0 minimum number and 100 maximum number so any number below 0 will make it 0 and above 100 will make it 100

# plt.hist(scores) for getting histogram chart

plt.hist(scores, bins=10,      # to make the number of bins (columns) equal to 10, you could count them to make sure
         color="lightgreen",   # to change th color of the histogram
         edgecolor="black") # to better find the edges we use edgecolor to change the histogram shape from one massive plob to a shape with nice outlines

plt.title("Exam Scores")
plt.xlabel("Score")
plt.ylabel("# of students")

plt.show()
```

```python
# subplots
import matplotlib.pyplot as plt
import numpy as np
# Figure  = The entire canvas
# AX = A single plot (subplot)

# print(plt.subplots(2, 2))
# we used a tuple 2 x 2 (2 rows & 2 columns) when you run this you will get the data for this subplot
# here it means the subplot will return a tuple, the tuple has two entities a figure object which is our canvas and a two dimensional array of axis objects
# when we call this function we will unpack the tuple right away
# rather than use print, we are going to assign what return to us and unpack it
# axes is technically a NumPy array

# figure, axes = plt.subplots(2, 2) # here we have four subplots 2 rows 2 columns
# figure, axes = plt.subplots(1, 2) # here we have two subplots 1 row 2 columns
# figure, axes = plt.subplots(2, 1) # here we have two subplots 2 rows 1 column
# figure, axes = plt.subplots(3, 3) # here we have nine subplots 3 rows 3 columns

x = np.array([1, 2, 3, 4, 5])



figure, axes = plt.subplots(2, 2)
# we will fill each of these individual subplots in

# we will start with the top left corner which is row [0] column [0] and we need some data so we back above and made varible x:
axes[0, 0].plot(x, x*2, color="red") # to make it simple we made y here equal to x multiplied by 2
axes[0, 0].set_title("X*2") # here to put a title for this subplot we should use the same indecies we are wotking on then use .set_title(" ")

# now for the right top subplot which is row [0] column [1]:
axes[0, 1].hist(x, x**2, color="blue") # we made the y axis here equal to x to the power of 2
axes[0, 1].set_title("X**2")

# now for the bottom left corner subplot which is row [1] column [0]
axes[1, 0].scatter(x, x**3, color="green") # we made the y axis here equal to x to the power of 3
axes[1, 0].set_title("X**3")

# now for the bottom right corner subplot which is row [1] column [1]
axes[1, 1].bar(x, x**4, color="purple") # we made the y axis here equal to x to the power of 4
axes[1, 1].set_title("X**4")

# you can use other shapes you don't need to stick to the plot shape for example you can use bar insted for any subplot
plt.tight_layout() # becuase some of the subplots are overlaping we use this code to fit them in the right place
plt.show()
```
