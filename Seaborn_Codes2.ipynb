**Distribution Plots:**

**1- Distribution Plot (distplot):**
   It's just going to provide a distribution for one variable.

**2- Joint Plot:**
  it's going to be used to compare two distribution and plots a scatter plot by default.

**3- KDE Plot (kernl density estimation):**
  It's used for the estimation of usage.

**4- Pair Plots:**
  It's used to plot the relationships across the entire data frames numerical values.

**5- Rug Plot:**
  It's used to plot a single column of data points in a dataframe as sticks so you see these different sticks across the bottom on this axis.

**Categoral Plots:**

Now with categorical plots we're going to focus on distributions using categorical data in reference to one of the pieces of new numerical data.

**1- Bar Plots:**
   to analyze some information between a category and the data giving the variance and by default this data is going to be aggregated based off of the maean

**2- Count Plot:**
   is kind a like bar plot but the estimator is just simply going to count the number of occurrences.

**3- Box Plot:**
   is going to allow you to compare different variables and what is going to do is to show you the quality of the data, the box is going to show quartiles of our data.

**4- Violin plot:**
   very similar to a box plot, Basically it is a combination between of a box plot as well as a kde plot.

**6- Strip Plot:**
   Basically what strip plot does is it's going to draw a scatter plot representing all the different data points where one variable is categorical

**7- Swarm Plot:**
  sort of combines the capabilities of multiple other plots, it's kind of like a strip plot but points are going to be adjusted so that they don't overlap and sorts of look like a combination between violin and strip plot whenever i create it.

```python
# Bar Plots:
# let's say let's go and let's analyze some information in regards to tipping for the different males and females
sns.barplot(x='sex', y='total_bill', data=tips_df)
# and here you can see the differences versus the sexes in regards to total bill from our tips data and this little bar right here is referencing the variance between the,
# males and the total bill amounts and so forth, and by default this data is going to be aggregated based off of the maean but you can come in and change that
```

```python
sns.barplot(x='sex', y='total_bill', data=tips_df, estimator= np.median) # we used NumPy function (median) here to change the default estimator
```

```python
sns.barplot(x='sex', y='total_bill', data=tips_df, estimator= np.std) # we used NumPy function (std: standard deviation) here to change the default estimator
```

```python
sns.barplot(x='sex', y='total_bill', data=tips_df, estimator= np.var) # we used NumPy function (var: variance) here to change the default estimator
```

```python
sns.barplot(x='sex', y='total_bill', data=tips_df, estimator= np.cov) # we used NumPy function (cov: covariance) here to change the default estimator
```

```python
# Count Plot:
# is kind a like bar plot but the estimator is just simply going to count the number of occurrences
# and how you call it:
sns.countplot(x='sex', data=tips_df) # to get the total number of males versus emailes in our dataset
```

```python
# Box Plot:
# is going to allow you to compare different variables and what is going to do is to show you the quality of the data
# lets see how men and women change in regards to the total bills over multiple different days
sns.boxplot(x='day', y='total_bill', data=tips_df, hue='sex')
# the box is going to show quartiles of our data see here is a legend and so forth, now this bar that goes across the middle is the median
# and the box is going to extend one standard deviation, that's what the box represents from our median
# these guys right here (T Shape) are called the whiskers and they're going to extend to all of the other data asides from what is in our standard deviation
# aside from those points that are considered outliers (o shape)

# Basically what hue is going to do for us is provide an additional way for us to add in another category
# so we're looking at the days as well as the total bills but by putting sex in here with hue that is also going to allow us to create multiple box plots for
# each of the days for the males and females so that we can compare them
# and we can see here that men tend to spend more money on friday versus on saturday where women spend more money

# and if you want to move the legend box because sometimes this legends gets in the way let's say you would want to move it you can go plot and legend and location
# this will normally almost always work if you just type in location zero

plt.legend(loc=0)
# plt.legend(loc='upper left') this is another way
```

```python
# Violin Plot:
# very similar to a box plot
# Basically it is a combination between of a box plot as well as a kde plot

sns.violinplot(x='day', y='total_bill', data=tips_df, hue='sex')
# now while a box plot is going to correspond to data points a violin plot uses the kde estimation of our data point to create these plots
```

```python
sns.violinplot(x='day', y='total_bill', data=tips_df, hue='sex', split=True)
# split is going to aloow you to compare how the categories compare to each other
```

```python
# Strip Plot:
# Basically what strip plot does is it's going to draw a scatter plot representing all the different data points where one variable is categorical
# and you're often going to see them used to show different observations with a box plot that is going to represent the average distributions

plt.figure(figsize=(8, 5))

sns.stripplot(x='day', y='total_bill', data=tips_df)
```

```python
# Cluster Map:
# we will divide the data based up on the different species of irises
iris = sns.load_dataset('iris')
species = iris.pop('species') # that's going to say how that we want to return values based off of the different species
sns.clustermap(iris)
# this is what we call a hierarchically clustered heat map, the right lines is the distance between the points are going to be calculated
# and the closest are going to be joined and this is going to continue for the next closest and it's going to then compare the columns and the rows of our heat map
```

```python
import pandas as pd
# what standard scale is going to do is: it will normalize the data to focus on the clustering which is going to sort of mess up the dates as you're going to see:

# Define the correct order for months
month_order = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
               'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

# Load the flights dataset again and convert 'month' to an ordered categorical type
flights_df_ordered = sns.load_dataset('flights')
flights_df_ordered['month'] = pd.Categorical(flights_df_ordered['month'], categories=month_order,
                                             ordered=True)

# Re-create the pivot table with the ordered month column
flights_pivot_ordered = flights_df_ordered.pivot_table(index='month', columns='year',
                                                       values='passengers')

sns.clustermap(flights_pivot_ordered, cmap='Blues', standard_scale=1)

# and there you can see our flight data, and you can also see that we really focused in and clustered our july's and august data, and you can also see down here that
# the years are no longer in order
# so the difference beatween the heat map and the cluster map is that:
# the heat map isn't going to try to reposition data to find clusters of data, it's going to sort of work with what it has
# while a cluster map is going to tightly fit and cluster the data, so you can see specific patterns that are being developed
```

```python
# Pair Grid:
# We talked before about pair plots and basically the way the pair plots works out is you plud in the data and it gives you a whole bunch of different plots
# but what happened whenever you want to get really specific control over what plots and what data shows up where
# well that's where pair grids come in

# I will create an empty grid system based off of the iris data
# and i will put a scatter plot across the upper lower and the diagonal

iris = sns.load_dataset('iris')
iris_g = sns.PairGrid(iris, hue='species')
iris_g.map(plt.scatter)
```

```python
iris = sns.load_dataset('iris')
iris_g = sns.PairGrid(iris, hue='species')
iris_g.map(plt.scatter)
iris_g.map_diag(plt.hist) # to put a histogram across the diagonal
```

```python
iris = sns.load_dataset('iris')
iris_g = sns.PairGrid(iris, hue='species')
iris_g.map_diag(plt.hist)
iris_g.map_offdiag(plt.scatter) # to have a scatter plots show off of the diagonal in the upper and the lower part
# and you can see now the histogram isn't mixed with scatter
```

```python
# let's say we want to go and put different stuff on the upper part and another different plot on the lower part
iris = sns.load_dataset('iris')
iris_g = sns.PairGrid(iris, hue='species')
iris_g.map_diag(plt.hist)

iris_g.map_upper(plt.scatter)
iris_g.map_lower(sns.kdeplot) # to put a kde plot on
```

```python
iris = sns.load_dataset('iris')

# we can define x and y for our custom grid:
# this is a method: iris_g = sns.PairGrid(iris, hue='species', vars=['sepal_length', 'sepal_width'])
# and this is other better method:
iris_g = sns.PairGrid(iris, hue='species',
                      x_vars=["sepal_length", "sepal_width"],
                      y_vars=["petal_length", "petal_width"])

iris_g.map(plt.scatter)
iris_g.add_legend()
```
