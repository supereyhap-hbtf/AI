```python
# Part 4 تقرير استكشاف داتاسيت تايتانيك (Titanic EDA Report)
import seaborn as sns
import pandas as pd

# تحميل الداتاسيت
titanic = sns.load_dataset('titanic')
print(titanic.head())

print("="*60)
print("      تقرير الاستكشاف الأولي لداتاسيت Titanic (Initial EDA)")
print("="*60)

# 1. نظرة عامة
rows, cols = titanic.shape
print(f"\n1. نظرة عامة على البيانات:")
print(f"• إجمالي الصفوف (الركاب): {rows}")
print(f"• إجمالي الأعمدة (المتغيرات): {cols}")
print("\n• شرح الأعمدة الرئيسية:")
print("  - survived : حالة النجاة (1 = ناجٍ، 0 = لم ينجُ)")
print("  - pclass   : درجة التذكرة / الدرجة الاجتماعية (1 = الأولى، 2 = الثانية، 3 = الثالثة)")
print("  - sex      : جنس الركب (male / female)")
print("  - age      : عمر الراكب بالسنوات")
print("  - fare     : سعر التذكرة المدفوع")

# 2. القيم المفقودة
missing = titanic.isnull().sum()
missing_filtered = missing[missing > 0]
print(f"\n2. القيم المفقودة بالأعمدة:")
print(missing_filtered)

# 3. الإحصاء الوصفي
avg_fare = titanic['fare'].mean()
max_age = titanic['age'].max()
min_age = titanic['age'].min()
print(f"\n قراءات إحصائية ملفتة:")
print(f"• متوسط سعر التذكرة: {avg_fare:.2f}")
print(f"• سن الراكب (أكبر عمر): {max_age} سنة")
print(f"• أصغر راكب (أقل عمر): {min_age} سنة (أقل من سنة واحدة)")

# 4. أسئلة استكشافية بالأكواد
overall_survival_rate = titanic['survived'].mean() * 100

female_survival_rate = titanic[titanic['sex'] == 'female']['survived'].mean() * 100
male_survival_rate = titanic[titanic['sex'] == 'male']['survived'].mean() * 100

pclass1_survival_rate = titanic[titanic['pclass'] == 1]['survived'].mean() * 100
pclass3_survival_rate = titanic[titanic['pclass'] == 3]['survived'].mean() * 100

avg_age_survived = titanic[titanic['survived'] == 1]['age'].mean()
avg_age_died = titanic[titanic['survived'] == 0]['age'].mean()

print(f"\n إجابات الأسئلة الاستكشافية:")
print(f"• نسبة الناجين الإجمالية: {overall_survival_rate:.2f}%")
print(f"• نسبة نجاة الإناث: {female_survival_rate:.2f}% | مقابل الذكور: {male_survival_rate:.2f}%")
print(f"• نسبة نجاة ركاب الدرجة الأولى: {pclass1_survival_rate:.2f}% | مقابل الدرجة الثالثة: {pclass3_survival_rate:.2f}%")
print(f"• متوسط عمر الناجين: {avg_age_survived:.2f} سنة | مقابل غير الناجين: {avg_age_died:.2f} سنة")

# Analytic Summary:
# خلاصة الاستكشاف الأولي (Titanic Key Findings):
# First Note:
# التفاوت في معدلات النجاة: بلغت نسبة النجاة بين الإناث حوالي 74.2% مقارنة بـ 18.9% فقط للذكور، مما يعكس إعطاء أولوية الإخلاء للنساء والأطفال
# Second Note:
# تأثير الفئة الاقتصادية (Class Effect):
# نجح 62.9% من ركاب الدرجة الأولى في النجاة، بينما لم ينجُ سوى 24.2% من ركاب الدرجة الثالثة، مما يدل على ارتباط وثيق بين موقع الكابينة/فئة التذكرة وفرص النجاة
# Third Note:
# الفقدان الكثيف في البيانات:
# يعاني عمود الكابينة (deck)
# من نقص حاد في البيانات بنسبة تتجاوز 77 %، يليه
# عمود العمر (age)
# بنقص 177 قيمة، مما يقتضي معالجة هذه القيم المفقودة قبل استخدام البيانات في بناء أي نماذج تنبؤية مستقبلاً

```
