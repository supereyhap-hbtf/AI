```python
#### الجزء 1 — اكتشاف ومعالجة القيم المفقودة
import seaborn as sns
import pandas as pd

# تحميل داتاسيت تايتانيك
titanic = sns.load_dataset('titanic')
df = pd.DataFrame(titanic)
print(df)
print(df.shape)

# 1. طباعة عدد القيم المفقودة بكل عمود مرتبة تنازلياً
missing_values = titanic.isnull().sum().sort_values(ascending=False)
print(" القيم المفقودة بكل عمود (مرتبة من الأكثر للأعددقل): ")
print(missing_values)
print("-" * 60)

# 2. التعويض في عمود
# age
# بالمتوسط وحفظه في عمود جديد
# age_filled
mean_age = titanic['age'].mean()
titanic['age_filled'] = titanic['age'].fillna(mean_age)

print("مقارنة ملخص العمر قبل وبعد التعويض: ")
print(f"المتوسط الذي تم التعويض به: {mean_age:.2f}")
print("عدد القيم المفقودة في age the original one :", titanic['age'].isnull().sum())
print("عدد القيم المفقودة في age_filled:", titanic['age_filled'].isnull().sum())
print("-" * 60)

# 3. القرار بشأن عمود deck
deck_missing_parts = (titanic['deck'].isnull().sum() / len(titanic)) * 100
print(f"نسبة القيم المفقودة في عمود deck هي: Percentege = {deck_missing_parts:.2f}%")
print("القرار وسببه:  ")
print("""
الأنسب هو حذف عمود 'deck' بالكامل
(باستخدام titanic.drop(columns=['deck'])).
السبب: لأن نسبة القيم المفقودة مرتفعة جداً وتتجاوز 77% من إجمالي البيانات، وبالتالي فإن أي تعويض
تخميني سيتسبب في تشويه البيانات، كما أن حذف الصفوف المفقودة سيفقدنا معظم حجم الداتاسيت.
""")
print("-" * 60)

# 4. التعويض في عمود embarked بالقيمة الأكثر تكراراً (Mode)
mode_embarked = titanic['embarked'].mode()[0]
titanic['embarked_filled'] = titanic['embarked'].fillna(mode_embarked)

print(f"القيمة الأكثر تكراراً في عمود embarked هي: '{mode_embarked}'")
print("عدد القيم المفقودة في embarked قبل التعديل:", titanic['embarked'].isnull().sum())
print("عدد القيم المفقودة في embarked بعد التعديل:", titanic['embarked_filled'].isnull().sum())
```

```python
#### الجزء 2 — اكتشاف وحذف التكرارات
import pandas as pd

# 1. إنشاء DataFrame لـ 6 عملاء مع تكرار صفين بالضبط
row_customers = pd.DataFrame({
    'customer_id': [101, 102, 103, 102, 104, 101],
    'name': ['إيهاب', 'سند', 'بسام', 'سند', 'ربى', 'إيهاب'],
    'city': ['عمان', 'إربد', 'الزرقاء', 'إربد', 'عمان', 'عمان']
})

print("جدول العملاء الأصلي (6 صفوف ببيانات مكررة): ")
print(row_customers)
print(f"عدد الصفوف الكلي: {len(row_customers)}")
print("-" * 60)

# 2. اكتشاف وطباعة الصفوف المكررة
duplicate_rows = row_customers[row_customers.duplicated()]
print("الصفوف المكررة التي تم اكتشافها بواسطة duplicated(): ")
print(duplicate_rows)
print("-" * 60)

# 3. حذف التكرارات مع الاحتفاظ بالظهور الأول (keep='first')
cleaned_customers = row_customers.drop_duplicates(keep='first')
print("الجدول بعد حذف التكرارات (drop_duplicates): ")
print(cleaned_customers)
print(f"عدد الصفوف الجديد: {len(cleaned_customers)} (تم حذف {len(row_customers) - len(cleaned_customers)} صفوف)")
```

```python
#### الجزء 3 — تلخيص البيانات بـ GroupBy
import seaborn as sns

titanic = sns.load_dataset('titanic')

# 1. متوسط سعر التذكرة (fare) لكل فئة ركاب (pclass)
fare_by_class = titanic.groupby('pclass')['fare'].mean()
print("متوسط سعر التذكرة لكل فئة ركاب: ")
print(fare_by_class)
print(f"--> الفئة الأكثر دفعاً بالمتوسط هي الفئة الأولى (pclass = 1) بمتوسط Average = {fare_by_class[1]:.2f} $")
print("-" * 60)

# 2. نسبة النجاة (survived) لكل جنس (sex)
survival_by_sex = titanic.groupby('sex')['survived'].mean()
print("نسبة النجاة حسب الجنس: ")
print(survival_by_sex)
print("-" * 60)

# 3. GroupBy : نسبة النجاة حسب الجنس والفئة معاً
survival_sex_class = titanic.groupby(['sex', 'pclass'])['survived'].mean()
print("نسبة النجاة حسب (الجنس + الفئة): ")
print(survival_sex_class)

max_group = survival_sex_class.idxmax()
max_val = survival_sex_class.max()
min_group = survival_sex_class.idxmin()
min_val = survival_sex_class.min()

print(f"\n• أعلى فئة + جنس في نسبة النجاة: {max_group} بنسبة percentege = {max_val*100:.1f}%")
print(f"• أقل فئة + جنس في نسبة النجاة : {min_group} بنسبة percentege = {min_val*100:.1f}%")
print("-" * 60)

# 4. عدد الركاب لكل ميناء صعدوا منه (embarked)
passengers_by_port = titanic.groupby('embarked')['passenger_id' if 'passenger_id' in titanic.columns else 'survived'].count()
print("عدد الركاب بكل ميناء صعود (S: Southampton, C: Cherbourg, Q: Queenstown):")
print(passengers_by_port)
```

```python
#### الجزء 4 — دمج الجداول: Merge, Join, Concat
import pandas as pd
import numpy as np

# ==========================================
# 1. تجهيز البيانات الأساسية
# ==========================================

# جدول العملاء
customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4],
    'name': ['أحمد', 'سارة', 'محمد', 'ليان'],
    'city': ['عمان', 'إربد', 'الزرقاء', 'عمان']
})

# جدول الطلبات (مع الملاحظة المتعمدة: العميل رقم 5 غير مسجل بجدول العملاء)
orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105],
    'customer_id': [1, 2, 1, 3, 5],
    'amount': [150, 200, 90, 300, 60]
})

print("--- 1. Inner Merge (التقاطع المشترك) ---")
# الدمج الداخلي: يظهر فقط العميل والطلب الموجودين معاً بالجدولين
inner_merged = pd.merge(customers, orders, on='customer_id', how='inner')
display(inner_merged)  # استخدام display لتنسيق جداول Colab التفاعلية
print("تفسير: طلب العميل 5 اختفى لأنه غير مسجل بجدول العملاء.\n")

print("--- 2. Left Merge (العملاء هم الأساس) ---")
# الدمج اليساري: يحتفظ بكل العملاء حتى لو لم يشتروا
left_merged = pd.merge(customers, orders, on='customer_id', how='left')
display(left_merged)
print("تفسير: العميل 4 (ليان) ظهرت بالجدول بـ NaN لعدم وجود طلبات لها.\n")

print("--- 3. Outer Merge (الدمج الكامل) ---")
# الدمج الخارجي الكامل: يحتفظ بكل شيء من الجدولين
outer_merged = pd.merge(customers, orders, on='customer_id', how='outer')
display(outer_merged)
print("""
تفسير الفروق بين الأنواع الثلاثة:
• Inner Merge: يعرض فقط القواطع المشتركة (العملاء 1, 2, 3 فقط).
• Left Merge : يحتفظ بجميع العملاء (1, 2, 3, 4)، والعميل بدون طلبات يأخذ قيم NaN.
• Outer Merge: يدمج كافة الصفوف؛ فيبقي العميل بدون طلبات (4) والطلب غير المعروف صاحبه (5).
""")

print("--- 4. الدمج الرأسي باستخدام pd.concat ---")
# جدول الطلبات الجديد لشهر فبراير
orders_feb = pd.DataFrame({
    'order_id': [106, 107],
    'customer_id': [2, 4],
    'amount': [180, 220]
})

# دمج الجدولين تحت بعض مع إعادة ضبط الفهرس لتفادي تكرار الأرقام
all_orders = pd.concat([orders, orders_feb], ignore_index=True)
display(all_orders)
```
